```python
marks = [78, 65, 90, 56, 72, 88, 65, 91, 45, 72, 80, 65, 55, 90, 72, 68, 88, 60, 75, 65]
#min marks
print(min(marks))
#max marks
print(max(marks))
#sum of marks
sum = 0
for i in range(0, len(marks)):
  sum = sum + marks[i]
print(sum)
#Average marks
n = len(marks)
mean = sum/n
print(mean)
#count of marks
print(len(marks))
#mode
import statistics as stats
print(stats.mode(marks))

c=stats.mode(marks)

print(c)

for i in range(0,len(marks),1):

  if(marks[i]==None):

    marks[i]=c

print(marks)
```
