```python
def countpass_st(marks):
  count=0
  count2=0
  for i in range(0,len(marks)):
    if(marks[i]>=50):
      count=count+1
    else:
      count2=count2+1
  return count,count2


l=[44,56,78,95,67,345,67,89,55]
c,d=countpass_st(l)
print("PASS",c)
print("Fail",d)


```
